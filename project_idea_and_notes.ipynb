#### VIRTUAL ENVIRONMENT used in this project.

##### * PROBLEM:
##### * Let's say you have a project and you deployed it on some computer "in the CLOUD".
##### * say : EQSRV - equiserver (to deploy)
##### * say : "Program A" is deployed which depends on python's "requests module 1.0 version"
##### * On this server there will also be a global python installation with all the libraries in it. (e.g.: "requests module 1.0 version")
##### * BUT If somebody deploy "Program B" 
##### * Let's say "Program B" requires requests 2.0 version.
##### * NOW someone will update the GLOBAL python library "requests" to version 2.0
##### * If someone do so, "Program A" might break/crash/show error.

##### * SOLUTION: Virtual Environment (Containerization)
##### * We provide virtualized containers to each Programs
##### * Everything that the program needs is present in that container (any library/module)
##### * So ypu don't have to look globally each time
##### * Outside/Globally if someone upgrade something, you're not affected
##### * Redudndant space is occupied to solve this problem (to not to break programs)

##### * For this install: uv python package manager for windows : https://docs.astral.sh/uv/
##### * once installed : "uv init" in vs code terminal 
##### * It'll initialize your uv python environment. You'll get a file "pyproject.toml" and ".python-version"
##### * run: "uv sync" : it'll install all the libraries locally in ".venv" (remember we used to do the same in nodejs?)
##### * we use "python --version" to check version globally
##### * for checking it locally/ & virtually -> go to cmd -> in respective folder run ".venv\Scripts\activate" -> run "python --version"
